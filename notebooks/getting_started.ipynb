## Data bundled with the package

```python
df = prenoms.load_example()
df.head()
```

## A static hvPlot figure

```python
import hvplot.pandas  # noqa: F401  (registers .hvplot on DataFrames)

df.hvplot.line(x="date", y="value", by="series", height=300, responsive=True)
```

## An interactive Panel widget (checks that Python <-> JS comms work)

```python
pn = prenoms.panel_extension()

series = pn.widgets.Select(name="series", options=sorted(df["series"].unique()))


def plot(s):
    return df[df["series"] == s].hvplot.line(x="date", y="value", height=250, responsive=True)


pn.Column(series, pn.bind(plot, series))
```
